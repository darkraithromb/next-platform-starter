```python
# app.py
import streamlit as st
import pandas as pd
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# NLTK downloads
nltk.download('punkt')
nltk.download('stopwords')

# Load data
essay_data = pd.read_csv("Training_Essay_Data.csv")  # Column: 'text'
sentiment_data = pd.read_csv("Sentiment_Analysis.csv")  # Column: 'sentiment'

# Merge and clean
data = pd.concat([essay_data, sentiment_data], axis=1)
data.dropna(inplace=True)

# Clean text
def clean_text(text):
    tokens = word_tokenize(text.lower())
    return " ".join([w for w in tokens if w not in stopwords.words("english") and w not in string.punctuation])

data['Cleaned'] = data['text'].apply(clean_text)

# Encode sentiment
le = LabelEncoder()
data['label'] = le.fit_transform(data['sentiment'])

# Split data
X_train, X_test, y_train, y_test = train_test_split(data['Cleaned'], data['label'], test_size=0.2, random_state=42)

# Train model
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=5000)),
    ('clf', RandomForestClassifier(n_estimators=100, random_state=42))
])
pipeline.fit(X_train, y_train)

# Humanizer function
def humanize(text):
    cleaned = clean_text(text)
    pred = pipeline.predict([cleaned])[0]
    sentiment = le.inverse_transform([pred])[0].lower()

    if sentiment in ["sadness", "worry", "anger", "hate"]:
        return f"Humanized (Empathetic): I understand — {text.strip().capitalize()}. It's okay to feel this way."
    elif sentiment in ["enthusiasm", "fun", "love", "happiness"]:
        return f"Humanized (Positive): Wow! That’s amazing! {text.strip().capitalize()} — Keep it up!"
    elif sentiment in ["relief", "surprise"]:
        return f"Humanized (Reflective): That's quite something! {text.strip().capitalize()}."
    else:
        return f"Humanized (Neutral): {text.strip().capitalize()}."

```

```python
%%writefile humanizer_app.py
import streamlit as st
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

nltk.download('punkt')
nltk.download('stopwords')

# Load data
essay_data = pd.read_csv("/content/Training_Essay_Data.csv")
sentiment_data = pd.read_csv("/content/Sentiment_Analysis.csv")

# Merge and clean
data = pd.concat([essay_data, sentiment_data], axis=1)
data.dropna(inplace=True)

def clean_text(text):
    text = text.lower()
    tokens = word_tokenize(text)
    return " ".join([w for w in tokens if w not in stopwords.words('english') and w not in string.punctuation])

data['Cleaned'] = data['text'].apply(clean_text)

label_encoder = LabelEncoder()
data['label'] = label_encoder.fit_transform(data['sentiment'])

X_train, X_test, y_train, y_test = train_test_split(data['Cleaned'], data['label'], test_size=0.2, random_state=42)

vectorizer = TfidfVectorizer(max_features=5000)
X_train_tfidf = vectorizer.fit_transform(X_train)
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_tfidf, y_train)

def predict_and_humanize(text):
    cleaned = clean_text(text)
    tfidf_input = vectorizer.transform([cleaned])
    pred = model.predict(tfidf_input)[0]
    sentiment = label_encoder.inverse_transform([pred])[0].lower()
    if sentiment in ["sadness", "worry", "anger", "hate"]:
        return f"Empathetic: I understand — {text.strip().capitalize()}. It's okay to feel this way."
    elif sentiment in ["enthusiasm", "fun", "love", "happiness"]:
        return f"Positive: Wow! That’s amazing! {text.strip().capitalize()} — Keep it up!"
    elif sentiment in ["relief", "surprise"]:
        return f"Reflective: That's quite something! {text.strip().capitalize()}."
    else:
        return f"Neutral: {text.strip().capitalize()}."

# Streamlit UI
st.title("🧠 AI Humanizer")
user_input = st.text_area("Enter text (AI-generated or plain):", height=150)

if st.button("Humanize"):
    if user_input.strip() != "":
        result = predict_and_humanize(user_input)
        st.subheader("Humanized Output")
        st.success(result)
```
